### 1. Draw a Shape Based on Input

In the next cell there's a `Text` widget where you can type "triangle" or "square" and an `IntSlider` where you can dial the size of the shape.

```python
shape = widgets.Text(description="Shape")
size = widgets.IntSlider(description="Size", min=50, max=200)
display(shape)
display(size)
```

Now, use the value from the widget and the turtle to draw a triangle or square depending on what's entered into the text box. If the box doesn't contain *either* the word "triangle" or "square" print an error message that states:

`Invalid shape`


```python
"""@inputshape""" 
```

```python
%%testing @inputshape as solution, shape, size
from contextlib import contextmanager
from unittest.mock import Mock
import ipywidgets as widgets 
import ast 
import re 
from nb_unittest import patch 

assert ast.If in solution.tokens, """I don't see an `if` in your answer."""
assert {"move", "turn", "print"} == solution.calls, "Your solution should only call tu.turn(), tu.move() and print()"""

for test_shape in [("triangle", 120, 66), ("square", 90, 77)]:
    shape.value = test_shape[0]
    size.value = test_shape[2]
    with patch('__main__.tu', feedback=f"While drawing a {test_shape[0]}...") as mock:
        solution.run()
        mock.turn.assert_called_with(test_shape[1])
        mock.move.assert_called_with(test_shape[2])    

shape.value = "bogus"
with patch('__main__.tu', feedback="With the word 'bogus' in the text box...") as mock:
    got = solution.run()
    mock.turn.assert_not_called()
    mock.move.assert_not_called()    
assert got.stdout.strip() == "Invalid shape", """Check the error message."""
```
