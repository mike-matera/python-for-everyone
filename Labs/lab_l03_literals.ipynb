### 2. String 

Write a **string** literal in the box below: 

```python
"""@str_cell"""
```

```python
%%testing @str_cell as solution
strs = [x for x in solution.constants if isinstance(x,str) 
        and x != "@str_cell".strip()]
assert len(strs) > 0, """I don't see a string."""
```

### 5. Scientific Notation
