# Literals

A literal is a specific value in your program. Literals get their name because they *literally* mean what they are. Numbers are literals in Python and so are words and phrases written in a human language. In this lesson you will learn how to communicate numbers and words to your program. You will also learn about *data types* in the Python language. 


A literal is not just something you type for fun. It is a value that your program can use immediately. When Python reads your program, it recognizes these literal values and uses them as the data for the next step. That is why literals are one of the most basic building blocks of any Python program.

There are different kinds of quotes. They are discussed in the Quotes and Formatting lesson. The most essential quotes are double quotes and single quotes and they are just two ways of doing the same thing. Copy these two examples into the cell below: 

```python
"Double Quote"
'Single Quote'
```

**Do you see any difference in the way Python prints the words?** There isn't! Double and single quotes do the same thing. Why have both? ## Data Types

So far we have seen literals that represent different kinds of values. In Python, each value belongs to a category called a *data type*. A data type tells Python how to interpret a value and what operations make sense for it.

Some common data types are:

- `int` for whole numbers, such as `10` or `-3`
- `float` for decimal numbers, such as `3.14` or `1e9`
- `str` for text, such as `"Hello"` or `"Python"`
- `bool` for true/false values, such as `True` or `False`
- `NoneType` for the special value `None`

You do not need to memorize every type right away. The important idea is that Python uses the literal you write to decide what kind of data it is working with. This helps Python know whether a value should be treated as a number, text, or a true/false condition.

Different data types support different operations. For example:

- Numbers can be added or subtracted: `10 + 5`
- Strings can be joined together: `"Hello" + " there"`
- Booleans can be used in decisions: `True and False`
- A string and a number cannot be added directly in the same way: `"Hello" + 5` would cause an error
