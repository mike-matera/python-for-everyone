## HTML and `display`

To Python, HTML is just a string. The Jupyter Notebook is what takes a Python string and uses your web browser to render the HTML in its beautiful form. In this lesson you'll use the `display` function,which works just like `print` except that `display` can show non-text output such as images, videos and HTML. The `display` function what the notebook uses to show the last expression in a cell.

Try running this display statement:

```python 
display("Hello World")
```

Now change the string to add the `<h1>` tag:

```python 
display("<h1>Hello World</h1>")
```

**Notice that it still just shows you the string!**. The `display` function needs to be told that the string contains HTML. That's why the `HTML` function exists. The `HTML` function converts a string into an `HTML` object. 

The `HTML` function needs to be imported:

````python
from IPython.display import HTML
````

Now try using the `HTML` function to convert the string:

```python
HTML("<h1>Hello World</h1>")
```

That's more like it! In the previous cell the `HTML` function returns the converted string and is on the last line of the cell, which is automatically displayed. If you want to have multiple HTML outputs displayed you can use `display` explicitly. For example:

```python
display(HTML("<h1>Hello World</h1>"))
display(HTML("<p>This is my Python notebook</p>"))
```

## Writing Python Code That Produces HTML 

Since HTML is just a string, programs that output HTML are similar in how they operate to programs that just `print` plain text. As a beginner, it might be a bit confusing at first trying to understand what parts of your program are Python and what parts are HTML, but it will get more clear with practice. 

Generating HTML with Python opens up some colorful possibilities. For example:

```python
color = 'green'
text = 'Hello World'
HTML(f'<span style="color: {color};"><b>{text}</b></span>')
```

**Notice the use of an f-string?**. Try changing the `color` and `text` variables in your code to see what happens. Mixing static and variable HTML is how every useful site works. It's also a place where the `format` function shines. 

```python
template = """
<span style="color: {};">
    <b>{}</b>
</span>
"""
HTML(template.format("orange", "I'm so happy!"))
```

Using format() this way enables your program to reuse the template to make new words and colors. Try running this code in the cell below and you'll see that the template can be used over and over:

```python
HTML(template.format("blue", "I'm blue"))
```

## Some HTML Example Programs

The Jupyter notebook makes it easy to experiment with HTML because it renders the HTML right in the notebook. Here are some example programs with HTML code that you can use and reuse:

````python
my_html = """
<h3>My Heading</h3>
<p>My first paragraph.</p>
"""
HTML(my_html)
````

### Heading Tags 

Heading tags change the size and weight of the letters. Words inside heading tags appear on their own line.

````python
my_html = """
<h1>Heading 1</h1>
<h2>Heading 2</h2>
<h3>Heading 3</h3>
<h4>Heading 4</h4>
<h5>Heading 5</h5>
"""
HTML(my_html)
````

### Text Decoration 

You can use tags to make bold, italic and underlined text. Notice that they don't split lines like heading tags do.

````python
my_html = """
This is <i>italic</i>.  
This is <b>bold</b>. 
This is <u>underlined</u>.
"""
HTML(my_html)
````

### Hyperlinks 

A hyperlink is a clickable reference to another page. There are two parts of a hyperlink, the link contents and the link target. The contents are what you click and the target is where you go when you click it. 

````python
my_html = """
Visit the 
<a href="http://www.lifealgorithmic.com">
    CIS-15
</a>
class page.
"""
HTML(my_html)
````

### Lists 

You can make ordered lists (lists with numbers) and unordered lists (lists with bullets) easily with HTML. These use nested elements.

````python
my_html = """
<ol>
    <li>First item</li>
    <li>Second item</li>
</ol>
"""
HTML(my_html)
````

````python
my_html = """
<ul>
    <li>Bullet one</li>
    <li>Bullet two</li>
</ul>
"""
HTML(my_html)
````

### Tables 

Tables let you organize elements into rows and columns. Tables nest just like lists. 

````python
my_html = """
<table>
    <tr>
        <th>Class</th>
        <th>Title</th>
    </tr>
    <tr>
        <td>cis-15</td>
        <td>Intro. to programming in Python</td>
    </tr>
    <tr>
        <td>cis-192</td>
        <td>UNIX/Linux Network Administration</td>
    </tr>
    <tr>
        <td>cis-194</td>
        <td>Windows Client Administration</td>
    </tr>
</table>
"""
HTML(my_html)
````

### Images 

You can display images in HTML using the image tag. 

````python
my_html = """
<img src="https://www.python.org/static/img/python-logo.png"/>
"""
HTML(my_html)
````

### Paragraphs and Breaks

You have seen how HTML "smashes" text together. You can make sure sentences stay separate using the paragraph `<p>` tag and the break `<br>` tag. 


````python
my_html = """ 
<p>Line one</p>
<p>Line two</p>
<p>Line three</p>
"""
HTML(my_html)
````

Paragraphs get extra space. Breaks are just a single "enter". 

````python
my_html = """
Line one<br/>
Line two<br/>
Line three<br/>
"""
HTML(my_html)
````
