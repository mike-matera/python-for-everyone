# Widgets! 

In this lesson you'll learn how your notebooks can take user input from widgets.

## What's a Widget?

What's a *widget*? A widget is a graphic on the screen that you click or type into. Buttons, menus and text boxes are familiar examples of widgets. You can embed widgets into a Jupyter notebook. That gives users of your notebook an easy way to enter values into your program. 

To use widgets in the notebook you have to `import` the Jupyter Widgets module:

````python
import ipywidgets as widgets 
````

Now you can create widgets that are intuitive to interact with:

````python
widgets.IntSlider(description="Pancakes", min=1, max=10)
````

The pancakes widget in the previous cell is fun but not useful because there's no way to get the number of pancakes into your code. To make widgets truly useful they should be assigned to a variable and shown in the notebook with `display`, just like `HTML` objects. 

This makes the slider value available to your program:

````python
pancakes = widgets.IntSlider(description="Pancakes", min=1, max=10)
display(pancakes)
````

Now pick a number of pancakes and run this code in the next cell:

```python
print("I want", pancakes.value, "pancakes")
```

````python
print("I want", pancakes.value, "pancakes")
````

The `pancakes.value` *attribute* contains the number of pancakes that is selected in the slider. Change the value an re-run the previous cell. Every time you run the cell it will show you the current pancake count. 

The trick works both ways! You can update the slider's value by assigning to the `pancakes.value` attribute. Type this code into the next cell:

```python
pancakes.value = 5
```

````python
pancakes.value = 5
````

**Notice that the slider jumps when you change the number!**

## Avaliable Widgets 

There are many widgets available and the list is growing. If you want to see them all, there's a complete [widget list](https://ipywidgets.readthedocs.io/en/stable/examples/Widget%20List.html) on the Jupuyter Widgets website. In this section I'll show you examples that we'll use in the class.

### String Widgets

If you want the user to provide you with a string, like their name, the `Text` and `Textarea` widgets provide a small and a large box for input:

````python
small_text = widgets.Text(
    description='String:',
    placeholder='Type something',
)
display(small_text)
````

Now get the value. Enter this code into the next cell:

```python
small_text.value
```

Here's how to get more text:

````python
big_text = widgets.Textarea(
    description='String:',
    placeholder='Type something',
)
display(big_text)
````

Now get the value typed in to `big_text`:

```python  
big_text.value
```

### Number Widgets

If you want the user of your notebook to enter a number you can use a string widget and convert the string to a number, but that is error prone. If you want the widget to make sure that a user inputs a valid number, there are sliders and boxes that do the job:

````python
number = widgets.IntSlider(
    description='Integer:',
    min=0,
    max=10,
    step=1,
)
display(number)
````

Now get the value. Notice that the value is of type `int`:

```python
print("value:", number.value, "type:", type(number.value))
```

Want a bigger range than a slider can provide? This lets your user type any integer:

````python
number = widgets.IntText(
    description='Integer:',
)
display(number)
````

**Notice how the box only accetps valid integers.** Check out the value:

```python 
print("value:", number.value, "type:", type(number.value))
```

The widgets package has floating point versions of the number widgets, too:

````python
number = widgets.FloatSlider(
    description='Float:',
    min=0,
    max=10.0,
    step=0.1,
)
````

Now notice that the type of `number.value` is `float`:

```python 
print("value:", number.value, "type:", type(number.value))
```

Just like with integers, you can make a box for input:

````python
number = widgets.FloatText(
    description='Float:',
)
display(number)
````

Now see the number that the user chose: 

```python
print("value:", number.value, "type:", type(number.value))
```

### Menus and Choices 

If you want the user of your notebook to have a fixed set of choices, you can create widgets that let them pick an item. 

````python
size = widgets.Dropdown(
    description='Size:',
    options=['Small', 'Medium', 'Large', 'X-Large'],
)
display(size)
````

Now `size.value` will always contain a valid choice. The user can't mess it up!

```python 
print("Your size is:", size.value)
```

If you want your user to see all their choices at once you can use radio buttons:

````python
topping = widgets.RadioButtons(
    description='Topping:',
    options=['Pepperoni', 'Pineapple', 'Anchovies'],
)
display(topping)
````

Now you know what topping was chosen:

```python
print("Pizza with:", topping.value)
```

### Complex Choices

There are kinds of inputs that are easy to talk about but that need complex widgets. Two examples of this are dates and colors. The calendar is complicated and there are so many colors to choose from. Here are two widgets that make it easy to pick a date and a color:

````python
event = widgets.DatePicker(
    description='Event:',
)
display(event)
````

The `date.value` you get from a `DatePicker` is a Python type that is used to hold dates in a program. It's very helpful for doing things like easily determining how many days, weeks or months ago an event was. 

```python
import datetime 

print("Event scheduled for:", event.value)

if event.value is not None:
    print(f"That's {event.value - datetime.date.today()} from now.")
```

The color picker is handy becuase it lets your user see what color they're picking. The value in `color.value` is a string that's usable in many places, including the Turtle.

````python
color = widgets.ColorPicker(
    description='Color',
)
display(color)
````

Now use the color with the turtle:

```python 
tu.set_color(color.value)
tu.move(100)
tu.turn(120)
tu.move(100)
tu.turn(120)
tu.move(100)
tu.turn(120)
```
