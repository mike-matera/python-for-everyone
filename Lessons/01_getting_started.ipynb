## Statements 

A *statement* is a single instruction that tells Python to do something. In a program, statements are the actions you want the computer to carry out, one after another. A statement may tell Python to display something, store a value, make a decision, or repeat an action. A `print` statement is one of the simplest examples. It tells Python to show a value to the reader. The `print` function is used inside a statement, but the statement itself is the whole instruction that tells Python what to do. In Python *all statements are expressions but not all expressions are statements*.

Here are examples of how to take the expressions from the previous section and make them into statements using `print`:

```python

print("Hello World")

```

### Functions

A function is a set of instructions that perform a task and has a name. Functions take input, called *arguments*, and may produce a result, called a *return value*. The arguments are written inside parentheses `(` and `)`, and multiple arguments are separated by commas. Some functions, like `print`, are used directly. Others are called on a value using a dot, such as `.upper()` or `.replace()`. A function can be used as part of a larger expression or as a complete statement.

Here are some examples of functions to try in the next cell. The `min` function looks for the smallest number of all of its arguments and returns it:

```python 
min(1, 2, 3, 4, 5)
```

The `len` function returns the length of its argument. If the argument is `"Hello World"`, it returns 11, which is the number of letters in "Hello World" plus one for the space.

```python 
len("Hello World")
```

Since the notebook displays the last statement, if you want to combine the last two examples and see both outputs you have to use the `print` function:

```python
print(sum(1, 2, 3, 4, 5))
print(len("Hello World"))
```

When Python evaluates a function call inside another function call, it works from the inside out. The innermost function is completed first, and its result is then used as the input for the next outer function. For example, in `print(len("Hello World"))`, Python first runs `len("Hello World")`, which returns 11, and then passes that result to `print`.

Enter each example in the next cell:

Functions like `min`, `len` and `print` are called *built-in* functions because they are always available and stand on their own. There are about 70 built-in function in Python. Another kind of function, called a *method*, requires a *subject*, that's placed before a dot (`.`). A method belongs to the subject and can sometimes be read like an English sentence. 

Here's an example of a method:

```python 
"i'm not shouting".upper()
```

You can read this statement as, *Take the message, "i'm not shouting," and convert it to upper case.*

Here's another example of a method. This one takes arguments:

```python 
"i'm not shouting".replace('not', 'really')
```

This statement can be read as, *Take the message, "i'm not shouting," and replace "not" with "really".*

Methods can be chained together to make code that is both compact and readable. When one method finishes, its result can be used as the input for the next method, allowing several transformations to be written in a single line.

Here is an example that combines the last two methods into one statement:

```python
"i'm not shouting".replace('not', 'really').upper()
```

### Compound Statements 

Later in the course we'll learn that some statements can *contain* other statements. These are known as *compound statements*. For now here's an example of a compound statement that prints a different greeting at different times of day. 

````python
from datetime import datetime
from zoneinfo import ZoneInfo

# Get the current clock time.
now = datetime.now(ZoneInfo("America/Los_Angeles"))  

# Compound statement: Decide what to say based on the current time
if now.hour < 12:
    print("Good morning.")
elif now.hour < 17:
    print("Good afternoon.")
else:
    print("Good evening.")
````
